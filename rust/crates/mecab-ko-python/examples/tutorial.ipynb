# MeCab-Ko Python Tutorial

This tutorial demonstrates how to use **mecab-ko-python** for Korean morphological analysis.

## Installation

```bash
pip install mecab-ko-python
```

## Features

- **morphs()**: Extract morphemes (형태소)
- **nouns()**: Extract nouns (명사)
- **pos()**: Part-of-speech tagging (품사 태깅)
- **parse()**: Full morphological analysis

Compatible with KoNLPy's Mecab interface!

## 1. Setup and Initialization

````python
from mecab_ko import Mecab

# Create tokenizer instance
mecab = Mecab()
print(f"MeCab tokenizer initialized: {mecab}")
````

## 2. Basic Usage: morphs()

Extract morphemes (형태소) - the smallest meaningful units in Korean.

````python
text = "안녕하세요"
morphs = mecab.morphs(text)
print(f"Text: {text}")
print(f"Morphemes: {morphs}")
````

````python
# More examples
examples = [
    "나는 학생입니다",
    "아버지가방에들어가신다",
    "자연어 처리는 재미있습니다"
]

for text in examples:
    morphs = mecab.morphs(text)
    print(f"Text: {text}")
    print(f"Morphemes: {morphs}")
    print()
````

## 3. Noun Extraction: nouns()

Extract only nouns (명사) from text - useful for keyword extraction and topic modeling.

````python
text = "인공지능과 자연어 처리 기술이 발전하고 있습니다"
nouns = mecab.nouns(text)
print(f"Text: {text}")
print(f"Nouns: {nouns}")
````

````python
# Compare morphs() vs nouns()
text = "아버지가방에들어가신다"
print(f"Text: {text}")
print(f"All morphemes: {mecab.morphs(text)}")
print(f"Only nouns: {mecab.nouns(text)}")
````

## 4. Part-of-Speech Tagging: pos()

Get morphemes with their POS (Part-of-Speech) tags.

### Common POS Tags:
- **NNG**: 일반 명사 (Common noun)
- **NNP**: 고유 명사 (Proper noun)
- **VV**: 동사 (Verb)
- **VA**: 형용사 (Adjective)
- **JX**: 보조사 (Auxiliary particle)
- **JKS**: 주격조사 (Subject particle)
- **EF**: 종결 어미 (Final ending)

````python
text = "나는 학생입니다"
pos_tags = mecab.pos(text)
print(f"Text: {text}")
print(f"POS Tags: {pos_tags}")
print()
print("Detailed breakdown:")
for morpheme, tag in pos_tags:
    print(f"  {morpheme:8s} -> {tag}")
````

````python
# Analyze a complex sentence
text = "서울대학교에서 자연어 처리 연구를 진행하고 있습니다"
pos_tags = mecab.pos(text)
print(f"Text: {text}")
print()
for morpheme, tag in pos_tags:
    print(f"  {morpheme:8s} -> {tag}")
````

## 5. Full Analysis: parse()

Get detailed morphological information in MeCab format.

````python
text = "형태소 분석"
parse_result = mecab.parse(text)
print(f"Text: {text}")
print("\nFull analysis:")
print(parse_result)
````

## 6. Practical Example: Word Frequency Analysis

Let's analyze a Korean text and visualize word frequencies.

````python
# Sample Korean text
text = """
인공지능은 현대 기술의 핵심입니다.
자연어 처리는 인공지능의 중요한 분야입니다.
형태소 분석은 자연어 처리의 기초 기술입니다.
한국어 형태소 분석기는 한국어 처리에 필수적입니다.
"""

# Extract nouns
nouns = mecab.nouns(text)
print(f"Total nouns extracted: {len(nouns)}")
print(f"Nouns: {nouns}")
````

````python
# Count word frequencies
from collections import Counter

noun_freq = Counter(nouns)
print("\nTop 10 most frequent nouns:")
for noun, count in noun_freq.most_common(10):
    print(f"  {noun:15s}: {count}")
````

## 7. Filtering by POS Tags

Extract specific types of words using POS tags.

````python
text = "아름다운 봄날 서울에서 친구를 만났습니다"
pos_tags = mecab.pos(text)

# Extract only verbs (VV)
verbs = [morpheme for morpheme, tag in pos_tags if tag.startswith('VV')]
print(f"Verbs: {verbs}")

# Extract only adjectives (VA)
adjectives = [morpheme for morpheme, tag in pos_tags if tag.startswith('VA')]
print(f"Adjectives: {adjectives}")

# Extract only proper nouns (NNP)
proper_nouns = [morpheme for morpheme, tag in pos_tags if tag == 'NNP']
print(f"Proper nouns: {proper_nouns}")
````

## 8. Batch Processing

Process multiple texts efficiently.

````python
texts = [
    "자연어 처리는 재미있습니다",
    "한국어는 아름다운 언어입니다",
    "형태소 분석기는 유용한 도구입니다",
    "파이썬으로 한국어를 처리할 수 있습니다"
]

# Process all texts
results = []
for text in texts:
    result = {
        'text': text,
        'morphs': mecab.morphs(text),
        'nouns': mecab.nouns(text),
        'pos_count': len(mecab.pos(text))
    }
    results.append(result)

# Display results
for i, result in enumerate(results, 1):
    print(f"\n{i}. {result['text']}")
    print(f"   Morphemes: {result['morphs']}")
    print(f"   Nouns: {result['nouns']}")
    print(f"   Token count: {result['pos_count']}")
````

## 9. Summary

### Key Methods:

| Method | Description | Returns |
